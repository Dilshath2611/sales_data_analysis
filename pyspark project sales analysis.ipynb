```python

from pyspark.sql.types import StructType, StructField, StringType, IntegerType, FloatType, DateType
sales_schema = StructType([
  StructField("product_id", IntegerType(), True),
  StructField("customer_id", StringType(), True),
  StructField("order_date", DateType(), True),
  StructField("location", StringType(), True),
  StructField("source_order", StringType(), True)
  ])

  
sales_df= spark.read.format('csv').option('inferSchema','true').schema(sales_schema).load('/Volumes/workspace/default/sales_data')
display(sales_df)
```

```python
from pyspark.sql.functions import month, year, quarter
sales_df = sales_df.withColumn('month', month('order_date')).withColumn('year', year('order_date')).withColumn('quarter', quarter('order_date'))
display(sales_df)
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, FloatType, DateType
sales_schema = StructType([
  StructField("product_id", IntegerType(), True),
  StructField("product_name", StringType(), True),
  StructField("price", StringType(), True)
  ])

  
menu_df= spark.read.format('csv').option('inferSchema','true').schema(sales_schema).load('/Volumes/workspace/default/menu/menu.csv.txt')
display(menu_df)
```

```python
total_amount_spent = sales_df.join(menu_df, sales_df.product_id == menu_df.product_id, 'inner').groupBy('customer_id').agg({'price':'sum'}).orderBy("customer_id")
display(total_amount_spent)
total_amount_spent = sales_df
```

```python
df1 = sales_df.join(menu_df, sales_df.product_id == menu_df.product_id, 'inner') \
    .groupBy('month').agg({'price': 'sum'}) \
    .orderBy("month")
display(df1)
```

```python
df2 = sales_df.join(menu_df, sales_df.product_id == menu_df.product_id, 'inner') \
    .groupBy('year').agg({'price': 'sum'}) \
    .orderBy("year")
display(df2)
```

```python
df3 = sales_df.join(menu_df, sales_df.product_id == menu_df.product_id, 'inner') \
    .groupBy('quarter').agg({'price': 'sum'}) \
    .orderBy("quarter")
display(df3)
```

```python
from pyspark.sql.functions import count
most_df = (sales_df.join(menu_df, 'product_id')
           .groupBy("product_id", "product_name")
           .agg(count('product_id').alias('product_count'))
           .orderBy('product_count', ascending=False)
           .drop("product_id"))

display(most_df)
```

```python
from pyspark.sql.functions import count
most_df = (sales_df.join(menu_df, 'product_id')
           .groupBy("product_id", "product_name")
           .agg(count('product_id').alias('product_count'))
           .orderBy('product_count', ascending=False)
           .drop("product_id").limit(5)
           )

display(most_df)
```

```python
from pyspark.sql.functions import count
most_df = (sales_df.join(menu_df, 'product_id')
           .groupBy("product_id", "product_name")
           .agg(count('product_id').alias('product_count'))
           .orderBy('product_count', ascending=False)
           .drop("product_id").limit(1)
           )

display(most_df)
```

```python
from pyspark.sql.functions import countDistinct
df= (sales_df.filter(sales_df.source_order =='Restaurant').groupBy('customer_id').agg(countDistinct('order_Date')))
display(df)
```

```python
total_sales_country = sales_df.join(menu_df, sales_df.product_id == menu_df.product_id, 'inner').groupBy('location').agg({'price':'sum'})
display(total_sales_country)
```
